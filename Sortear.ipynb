# **PROGRAMA PARA SIRTEAR 6 NÚMEROS ENTRE 1 E 60



## ALGORITMO PARA "SORTEIO DE NUMEROS":


```python
VAR
  numeros_sorteados: lista de inteiros
  numero: inteiro

INÍCIO:
  numeros_sorteados <- sortear 6 números diferentes entre 1 e 60

  ordenar numeros_sorteados em ordem crescente

  Escrever "Números sorteados:"

  PARA CADA numero EM numeros_sorteados FAÇA:

    escrever numero

  FIM_PARA
FIM
```

```python
import random
numeros_sorteados = random.sample(range(1, 61), 6)
numeros_sorteados.sort()
print("Números sorteados:")
for numero in numeros_sorteados:
    print(numero, end = " ")
```
